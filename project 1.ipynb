```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
import joblib
import os
```

```python
data = pd.read_csv("polar_energy_dataset.csv")
data
```

```python
x= data[["temperature_c","wind_speed_mps","solar_radiation_wm2","battery_level_percent","hour","solar_power_kwh","wind_power_kwh"]]
x
```

```python
y = data["energy_demand_kwh"]
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = RandomForestRegressor(n_estimators=200,random_state=42)
model.fit(x_train, y_train)
predictions = model.predict(x_test)
```

```python
mae = mean_absolute_error(y_test, predictions)
print("MAE:", round(mae, 2))
```

```python
r2 = r2_score(y_test, predictions)
print("r2:",round(r2,2))
```

```python
def optimize_energy(demand,solar_power,wind_power,battery_level):
  renewable_energy = solar_power + wind_power
  result = {}
  if renewable_energy >= demand:
     result["source"] = "Solar + Wind"
     result["battery_action"] = "Charge"
     result["generator"] = "OFF"
     result["message"] = ("Renewable energy is sufficient and Store extra energy in the battery.")

   
  elif battery_level > 30:

        result["source"] = "Renewable + Battery"

        result["battery_action"] = "Discharge"

        result["generator"] = "OFF"

        result["message"] = ("Use renewable energy first and cover the remaining demand using the battery.")

   
  else:
     result["source"] = "Renewable + Generator"
     result["battery_action"] = "Preserve"
     result["generator"] = "ON"
     result["message"] = ("Battery level is low Activate backup generator.")
  return result

result = optimize_energy(
    demand=150,
    solar_power=40,
    wind_power=60,
    battery_level=70)

print(result)
```
